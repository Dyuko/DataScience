### Pregunta 6

¿Cuales son las 10 palabras mas frecuentes (sin tener en cuenta *Stopwords*) en los documentos que no son spam y spam?


*Esta función debe devolver una tupla (palabras mas frecuentes, no spam, palabras mas frecuentes, spam).*

```python
def pregunta_seis():
    # Filtrar textos de no spam y spam usando las etiquetas (y_train)
    non_spam_texts = X_train[y_train == 0]
    spam_texts = X_train[y_train == 1]

    # Vectorizador para no spam
    vectorizer_non_spam = CountVectorizer(stop_words='english')
    X_non_spam = vectorizer_non_spam.fit_transform(non_spam_texts)
    non_spam_counts = np.asarray(X_non_spam.sum(axis=0)).flatten()
    top_non_spam_indices = np.argsort(non_spam_counts)[-10:][::-1]
    top_non_spam_words = vectorizer_non_spam.get_feature_names_out()[top_non_spam_indices]

    # Vectorizador para spam
    vectorizer_spam = CountVectorizer(stop_words='english')
    X_spam = vectorizer_spam.fit_transform(spam_texts)
    spam_counts = np.asarray(X_spam.sum(axis=0)).flatten()
    top_spam_indices = np.argsort(spam_counts)[-10:][::-1]
    top_spam_words = vectorizer_spam.get_feature_names_out()[top_spam_indices]

    return (list(top_non_spam_words), list(top_spam_words))

pregunta_seis()
```

### Pregunta 7

¿Cuales son las 10 palabras mas frecuentes (solo teniendo en cuenta *Stopwords*) en los documentos que no son spam y spam?


*Esta función debe devolver una tupla (palabras mas frecuentes, no spam, palabras mas frecuentes, spam).*

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
def pregunta_siete():
    # Filtrar textos de no spam y spam usando las etiquetas (y_train)
    non_spam_texts = X_train[y_train == 0]
    spam_texts = X_train[y_train == 1]

    # Vectorizador que INCLUYE TODAS LAS PALABRAS (sin eliminar stopwords)
    vectorizer = CountVectorizer()

    # Procesar no spam
    X_non_spam = vectorizer.fit_transform(non_spam_texts)
    non_spam_counts = np.asarray(X_non_spam.sum(axis=0)).flatten()
    feature_names = vectorizer.get_feature_names_out()

    # Filtrar solo stopwords y obtener las 10 más frecuentes
    stopword_indices = [i for i, word in enumerate(feature_names) if word in ENGLISH_STOP_WORDS]
    non_spam_stop_counts = non_spam_counts[stopword_indices]
    top_non_spam_stop_indices = np.argsort(non_spam_stop_counts)[-10:][::-1]
    top_non_spam_stopwords = feature_names[stopword_indices][top_non_spam_stop_indices]

    # Procesar spam
    X_spam = vectorizer.fit_transform(spam_texts)
    spam_counts = np.asarray(X_spam.sum(axis=0)).flatten()
    feature_names = vectorizer.get_feature_names_out()

    # Filtrar solo stopwords y obtener las 10 más frecuentes
    stopword_indices = [i for i, word in enumerate(feature_names) if word in ENGLISH_STOP_WORDS]
    spam_stop_counts = spam_counts[stopword_indices]
    top_spam_stop_indices = np.argsort(spam_stop_counts)[-10:][::-1]
    top_spam_stopwords = feature_names[stopword_indices][top_spam_stop_indices]

    return (list(top_non_spam_stopwords), list(top_spam_stopwords))

pregunta_siete()
```


La siguiente función te ayudará a combinar nuevas caracteristicas en tus datos de entrenamiento:

```python
def add_feature(X, feature_to_add):
    """
    Returns sparse feature matrix with added feature.
    feature_to_add can also be a list of features.
    """
    from scipy.sparse import csr_matrix, hstack
    return hstack([X, csr_matrix(feature_to_add).T], 'csr')
```

### Pregunta 11

Ajustar y transformar los datos de entrenamiento `X_train` usando un `TfidfVectorizer` ignorando los términos que tienen una frecuencia de documento estrictamente inferior a **5**.

Usando esta matriz de término de documento y una característica adicional, **la longitud del documento (número de caracteres)**, ajustar a un modelo de Clasificación de Vector de Soporte con regularización `C = 10000`. Luego calcule el área bajo de la curva (AUC) usando los datos de prueba transformados.

*Esta función debe devolver el AUC como un flotante.*

```python
from sklearn.svm import SVC

def respuesta_once():
    # 1. Vectorización TF-IDF con términos que aparecen en al menos 5 documentos
    vectorizer = TfidfVectorizer(min_df=5).fit(X_train)
    X_train_tfidf = vectorizer.transform(X_train)
    X_test_tfidf = vectorizer.transform(X_test)

    # 2. Añadir característica adicional: longitud del documento (número de caracteres)
    X_train_len = X_train.str.len().values
    X_test_len = X_test.str.len().values

    X_train_combined = add_feature(X_train_tfidf, X_train_len)
    X_test_combined = add_feature(X_test_tfidf, X_test_len)

    # 3. Entrenar modelo SVM con C=10000
    svm = SVC(C=10000, probability=True)
    svm.fit(X_train_combined, y_train)

    # 4. Calcular AUC en los datos de prueba
    y_pred_proba = svm.predict_proba(X_test_combined)[:, 1]
    auc = roc_auc_score(y_test, y_pred_proba)

    return float(auc)
```

### Pregunta 12

Ajustar y transformar los datos de entrenamiento `X_train` usando un `TfidfVectorizer` ignorando los términos que tienen una frecuencia de documento estrictamente inferior a **5** y usando **n-gramas de palabras n = 1 a n = 3** (unigramas, bigramas y trigramas).

Usando esta matriz de término-documento y las siguientes características adicionales:
* la longitud del documento (número de caracteres)
* **cantidad de dígitos por documento**

Ajustar un modelo de Regresión logística con regularización `C = 100`. Luego calcule el área bajo de la curva (AUC) usando los datos de prueba transformados.

*Esta función debe devolver el AUC como un flotante.*

```python
from sklearn.linear_model import LogisticRegression

def respuesta_doce():
    # 1. Vectorización TF-IDF con n-gramas (1-3) y min_df=5
    vectorizer = TfidfVectorizer(ngram_range=(1, 3), min_df=5).fit(X_train)
    X_train_tfidf = vectorizer.transform(X_train)
    X_test_tfidf = vectorizer.transform(X_test)

    # 2. Calcular características adicionales
    # Longitud del documento (número de caracteres)
    X_train_len = X_train.str.len()
    X_test_len = X_test.str.len()

    # Cantidad de dígitos por documento
    X_train_digits = X_train.str.count(r'\d')
    X_test_digits = X_test.str.count(r'\d')

    # 3. Combinar características usando add_feature
    # Primero añadimos la longitud
    X_train_combined = add_feature(X_train_tfidf, X_train_len)
    X_test_combined = add_feature(X_test_tfidf, X_test_len)

    # Luego añadimos el conteo de dígitos
    X_train_combined = add_feature(X_train_combined, X_train_digits)
    X_test_combined = add_feature(X_test_combined, X_test_digits)

    # 4. Entrenar modelo de Regresión Logística
    lr = LogisticRegression(C=100, max_iter=1000)
    lr.fit(X_train_combined, y_train)

    # 5. Calcular AUC en los datos de prueba
    y_pred_proba = lr.predict_proba(X_test_combined)[:, 1]
    auc = roc_auc_score(y_test, y_pred_proba)

    return float(auc)
```

```python
respuesta_doce()
```

### Pregunta 13

Ajustar y transformar los datos de entrenamiento `X_train` usando un `CountVectorizer` ignorando los términos que tienen una frecuencia de documento estrictamente inferior a **5** y utilizando **caracteres n-grams desde n = 2 a n = 5.**

Para decirle a `CountVectorizer` que use caracteres n-grams, pase en `analyzer = 'char_wb'` que crea caracteres n-gramas solo del texto dentro de los límites de las palabras. Esto debería hacer que el modelo sea más robusto a los errores ortográficos.

Usando esta matriz término documento y las siguientes características adicionales:
* la longitud del documento (número de caracteres)
* cantidad de dígitos por documento
* **cantidad de caracteres que no son palabras (cualquier cosa que no sea una letra, dígito o guión bajo).**

Ajustar un modelo de Regresión logística con regularización `C = 100`. Luego calcule el área bajo de la curva (AUC) usando los datos de prueba transformados.

También **encuentre los 10 coeficientes más pequeños y los 10 más grandes del modelo** y devuélvalos junto con el AUC en una tupla.

La lista de los 10 coeficientes más pequeños debe ordenarse primero, la lista de los 10 coeficientes más grandes debe ordenarse primero.

Las tres características que se agregaron a la matriz de términos del documento deben tener los siguientes nombres si aparecen en la lista de coeficientes:
['longitud_doc', 'conteo_digito', 'caracteres_no_palabra']

*Esta función debe devolver una tupla `(AUC como flotante, lista de coeficientes más pequeños, lista de coeficientes más grande)`.*

```python
def respuesta_trece():
    # 1. Vectorización con CountVectorizer (caracteres n-grams 2-5)
    vectorizer = CountVectorizer(ngram_range=(2, 5), analyzer='char_wb', min_df=5).fit(X_train)
    X_train_vec = vectorizer.transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    # 2. Calcular características adicionales
    # Longitud del documento
    X_train_len = X_train.str.len()
    X_test_len = X_test.str.len()

    # Cantidad de dígitos
    X_train_digits = X_train.str.count(r'\d')
    X_test_digits = X_test.str.count(r'\d')

    # Caracteres no palabras (no letras, dígitos o _)
    X_train_non_word = X_train.str.count(r'[^\w]')
    X_test_non_word = X_test.str.count(r'[^\w]')

    # 3. Combinar características usando add_feature
    X_train_combined = add_feature(X_train_vec, X_train_len)
    X_train_combined = add_feature(X_train_combined, X_train_digits)
    X_train_combined = add_feature(X_train_combined, X_train_non_word)

    X_test_combined = add_feature(X_test_vec, X_test_len)
    X_test_combined = add_feature(X_test_combined, X_test_digits)
    X_test_combined = add_feature(X_test_combined, X_test_non_word)

    # 4. Entrenar modelo de Regresión Logística
    lr = LogisticRegression(C=100, max_iter=1000)
    lr.fit(X_train_combined, y_train)

    # 5. Calcular AUC
    y_pred_proba = lr.predict_proba(X_test_combined)[:, 1]
    auc = roc_auc_score(y_test, y_pred_proba)

    # 6. Obtener coeficientes
    feature_names = np.concatenate([
        vectorizer.get_feature_names_out(),
        ['longitud_doc', 'conteo_digito', 'caracteres_no_palabra']
    ])

    coef_df = pd.DataFrame({
        'feature': feature_names,
        'coef': lr.coef_[0]
    })

    # Coeficientes más pequeños (10 más negativos)
    smallest_coef = (
        coef_df.nsmallest(10, 'coef')
        .sort_values('coef')
        [['feature', 'coef']]
        .apply(tuple, axis=1)
        .tolist()
    )

    # Coeficientes más grandes (10 más positivos)
    largest_coef = (
        coef_df.nlargest(10, 'coef')
        .sort_values('coef', ascending=False)
        [['feature', 'coef']]
        .apply(tuple, axis=1)
        .tolist()
    )

    return (float(auc), smallest_coef, largest_coef)
```
